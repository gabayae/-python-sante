## 1. Les trois mécanismes de données manquantes

Tout traitement des valeurs manquantes depend de leur **mécanisme** (Rubin 1976) :

- **MCAR — Missing Completely At Random.** Independant du reste. Exemple : un tube de prelevement casse. Supprimer est sans biais.
- **MAR — Missing At Random.** Depend de variables **observees**. Exemple : le cholesterol n'est pas mesure chez les jeunes. Imputer en tenant compte de l'age est sans biais.
- **MNAR — Missing Not At Random.** Depend de la valeur **non observee**. Exemple : les patients les plus malades ratent leurs rendez-vous. Supprimer biaise vers les patients en bonne santé.

> Dans les dossiers electroniques, MNAR est la **règle**, pas l'exception. Toujours vérifier si l'issue differe entre lignes completes et lignes a données manquantes.

Lecture recommandée : Little RJA & Rubin DB, *Statistical Analysis with Missing Data*, 3e ed., Wiley 2019.

## 3. Strategies d'imputation

| Methode             | Quand l'utiliser                                          |
|---------------------|-----------------------------------------------------------|
| **Mediane**         | Variables continues asymétriques (lab, tension, IMC)      |
| **Moyenne**         | Variables continues symétriques (taille adulte)            |
| **Mode**            | Variables catégorielles                                   |
| **Par sous-groupe** | Bonne valeur dependant d'un groupe connu (sexe, age)      |
| **KNN**             | Plusieurs colonnes correlees avec manquements partages    |

Toujours **documenter** la méthode utilisee. Pour les variables critiques, garder un **indicateur de manquement** avant d'imputer.

> **Syntaxe — fillna**
> ```python
> df["col"] = df["col"].fillna(valeur)             # remplace par valeur fixe
> df["col"] = df["col"].fillna(df["col"].median())   # par mediane
> df = df.fillna({"a": 0, "b": "inconnu"})        # plusieurs colonnes
> df = df.copy()                                   # éviter SettingWithCopyWarning
> ```

> **Syntaxe — dropna**
> ```python
> df.dropna()                              # supprime toute ligne avec NaN
> df.dropna(subset=["col1", "col2"])       # seulement si NaN dans ces colonnes
> df.dropna(thresh=3)                      # garde lignes avec >= 3 non-NaN
> ```

````python
df_imp = df.copy()                             # toujours travailler sur une copie

# Mediane pour les valeurs lab (robuste aux extremes)
df_imp["tension"] = df_imp["tension"].fillna(df_imp["tension"].median())
df_imp["hba1c"] = df_imp["hba1c"].fillna(df_imp["hba1c"].median())

# Moyenne pour l'age (a peu près symétrique dans cette cohorte adulte)
df_imp["age"] = df_imp["age"].fillna(df_imp["age"].mean())

# Mode pour la catégorielle
# .mode() peut renvoyer plusieurs valeurs si égalité, on prend la 1ere : [0]
df_imp["sexe"] = df_imp["sexe"].fillna(df_imp["sexe"].mode()[0])
df_imp
````

## 4. Validation : plages cliniquement plausibles

On enumere les bornes attendues dans un **dictionnaire** (`col -> (min, max)`) puis on les applique systematiquement.

> **Syntaxe — definition de fonction avec docstring**
> ```python
> def ma_fonction(arg1, arg2):
>     """Une ligne qui resume."""
>     # ... corps ...
>     return résultat
> ```

````python
# Dictionnaire de plages cliniquement acceptables
PLAGES = {
    "age":     (0, 120),         # tuple (min, max)
    "tension": (60, 300),
    "hba1c":   (2.0, 20.0),
}

def valider_plage(df, colonne, lo, hi):
    """Renvoie le sous-DataFrame des lignes hors plage [lo, hi]."""
    # Operateur | == OR booleen vectorise pour pandas
    hors = df[(df[colonne] < lo) | (df[colonne] > hi)]
    if len(hors):
        print(f"  '{colonne}' : {len(hors)} valeur(s) hors [{lo}, {hi}]")
    return hors

# .items() : iterer sur (cle, valeur) d'un dictionnaire
# (lo, hi) : 'unpacking' d'un tuple en 2 variables
for col, (lo, hi) in PLAGES.items():
    if col in df_imp.columns:           # seulement si la colonne existe
        valider_plage(df_imp, col, lo, hi)
print("Validation terminee.")
````

## 5. Pipeline de nettoyage reutilisable

On combine tout dans une seule fonction. Patron : chaque étape prend un DataFrame et en renvoie un. C'est testable, debuggable et explicite sur chaque choix.

> **Syntaxe — argument par défaut**
> ```python
> def f(x, y=10):       # y vaut 10 si non spécifie
>     ...
> f(5)                  # appel avec x=5, y=10
> f(5, y=20)            # appel avec x=5, y=20
> ```

````python
def nettoyer_clinique(df, plages=PLAGES):
    """Pipeline de base : noms, doublons, plages, imputation.

    Parametres
    ----------
    df : pd.DataFrame
        Données brutes.
    plages : dict[str, tuple[float, float]]
        Bornes attendues pour chaque colonne numérique.
    """
    df = df.copy()                       # toujours une copie

    # ---- 1) Noms de colonnes propres ----
    # .str.strip() supprime les espaces aux extremites
    # .str.lower() passe en minuscules
    # .str.replace(" ", "_") remplace espaces par underscores
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

    # ---- 2) Supprimer les doublons exacts ----
    avant = len(df)
    df = df.drop_duplicates()
    print(f"  doublons supprimes : {avant - len(df)}")

    # ---- 3) Hors plage -> NaN ----
    for col, (lo, hi) in plages.items():
        if col in df.columns:
            masque = (df[col] < lo) | (df[col] > hi)
            if masque.any():
                print(f"  '{col}' hors plage -> NaN : {int(masque.sum())}")
                # df.loc[ligne, colonne] = valeur
                df.loc[masque, col] = np.nan

    # ---- 4) Imputation mediane sur les colonnes numériques ----
    # .select_dtypes(include="number") garde seulement int/float
    for col in df.select_dtypes(include="number").columns:
        df[col] = df[col].fillna(df[col].median())

    return df


# Demo sur un DataFrame deliberement sale
messy = pd.DataFrame({
    "AGE":     [45, 67, -3, 52, 200, 38],            # -3 et 200 invalides
    "tension": [128, 158, 112, 450, 135, np.nan],   # 450 invalide, 1 NaN
    "hba1c":   [5.4, 7.8, 5.1, 6.3, np.nan, 5.8],
})
nettoyer_clinique(messy)
````

## 6. Tendance centrale — moyenne, mediane, mode

| Mesure   | Quand l'utiliser                                |
|----------|-------------------------------------------------|
| Moyenne  | Distribution approximativement symétrique       |
| Mediane  | Distribution asymétrique (lab, duree de sejour) |
| Mode     | Variable catégorielle                           |

**Regle pratique.** Si une valeur extreme deplace la moyenne de plus de quelques pourcents, **rapporter la mediane**.

> **Syntaxe — Series pandas**
> ```python
> s = pd.Series([v1, v2, v3])      # liste -> Series (vecteur 1D)
> s.mean(), s.median(), s.std()    # moyenne, mediane, écart-type
> s.drop(index)                    # supprime un element par index
> ```

````python
# Glycemie a jeun avec une valeur aberrante (saisie erronee : 340)
glycémie = pd.Series([92, 98, 104, 95, 88, 110, 97, 340, 101, 93])

# Comparer moyenne et mediane montre la sensibilité a l'outlier
print(f"Moyenne : {glycémie.mean():.1f}  <- tiree par 340")
print(f"Mediane : {glycémie.median():.1f}  <- robuste")

# .drop(7) supprime l'element a l'index 7 (la valeur 340)
print(f"Sans la valeur aberrante : moyenne = {glycémie.drop(7).mean():.1f}")
````

## 7. Dispersion — écart-type, IQR, CV

L'**IQR** (Q3 - Q1) est la dispersion des 50 % du milieu. Robuste aux valeurs extremes, c'est le bon résumé pour les distributions asymétriques (duree de sejour, couts hospitaliers).

Le **coefficient de variation** $CV = \sigma / \mu$ (en %) est sans unite : utile pour comparer la dispersion de variables aux unites différentes.

> **Syntaxe — quantile**
> ```python
> s.quantile(0.5)                  # mediane (50e percentile)
> s.quantile([0.25, 0.75])         # plusieurs en même temps
> q1, q3 = s.quantile([0.25, 0.75])  # unpacking
> ```

````python
# Duree moyenne de sejour (DMS), distribution asymétrique a droite
dms = pd.Series([2, 3, 3, 4, 4, 5, 5, 6, 7, 8, 12, 28, 45])

# Recuperation des quartiles en une ligne (unpacking d'une Series)
q1, q3 = dms.quantile([0.25, 0.75])
print(f"Mediane : {dms.median():.0f} jours")
print(f"IQR     : {q1:.0f}-{q3:.0f} jours")
print(f"Moyenne : {dms.mean():.1f} jours  <- gonflee par 28 et 45")
````

## 8. Mesures épidémiologiques

### 8.1 Prevalence

$$P = \frac{\text{cas existants au temps }t}{\text{population au temps }t}$$

Toujours accompagner d'un **intervalle de confiance**. L'approximation normale convient quand $n p (1-p) > 5$.

> **Syntaxe — math.sqrt et puissance**
> ```python
> math.sqrt(x)        # racine carree
> x ** 0.5            # équivalent
> x ** 2              # x au carre
> ```

````python
# Données : 625 diabétiques sur 5,000 enquetes
n_total, n_diabetique = 5_000, 625    # _ comme separateur de milliers

p = n_diabetique / n_total            # proportion brute
se = math.sqrt(p * (1 - p) / n_total) # erreur standard (formule binomiale)

# IC 95% normal : p +/- 1.96 * SE
ci_lo = (p - 1.96 * se) * 100         # en pourcentage
ci_hi = (p + 1.96 * se) * 100

print(f"Prevalence du diabete : {p * 100:.1f}%   IC 95% [{ci_lo:.1f}%, {ci_hi:.1f}%]")
````

### 8.2 Taux d'incidence (nouveaux cas par personne-temps)

$$IR = \frac{\text{nouveaux cas pendant la période}}{\text{personnes-années suivies}}$$

Les nouveaux cas suivent une loi de Poisson, d'ou $\text{SE}(IR) = \sqrt{\text{cas}} / \text{PA}$.

````python
# Cohorte TB : 18 nouveaux cas, 850 personnes-années suivies
nouveaux_tb, personne_annees = 18, 850
ir = nouveaux_tb / personne_annees
se = math.sqrt(nouveaux_tb) / personne_annees    # SE Poisson

# On multiplie par 1000 pour une lecture habituelle 'pour 1,000 PA'
ci_lo, ci_hi = (ir - 1.96 * se) * 1000, (ir + 1.96 * se) * 1000
print(f"Incidence TB : {ir * 1000:.1f} pour 1,000 PA")
print(f"IC 95%       : [{ci_lo:.1f}, {ci_hi:.1f}]")
````

### 8.3 Ratio de risque (RR) et ratio de cotes (OR)

Tableau 2x2 :

|              | Malade + | Malade - |
|--------------|----------|----------|
| **Expose**   | a        | b        |
| **Non expose** | c      | d        |

$$RR = \frac{a/(a+b)}{c/(c+d)} \qquad OR = \frac{a \cdot d}{b \cdot c}$$

On passe par $\ln$ pour les IC (la distribution de $\ln RR$ et $\ln OR$ est approximativement normale).

> **Syntaxe — fonction qui renvoie un dictionnaire**
> ```python
> def f():
>     return {"résultat": x, "ic": (lo, hi)}    # dict
> r = f()
> r["résultat"], r["ic"]
> ```

````python
def ratio_risque(a, b, c, d, alpha=0.05):
    """Risk ratio avec IC Wald sur l'échelle log."""
    # stats.norm.ppf(p) = quantile p de la loi normale standard
    z = stats.norm.ppf(1 - alpha / 2)         # ~1.96 pour alpha=0.05
    rr = (a / (a + b)) / (c / (c + d))
    se = math.sqrt(1/a - 1/(a+b) + 1/c - 1/(c+d))
    return {
        "RR":    round(rr, 3),
        # math.exp() = e^x, inverse de math.log()
        "IC_lo": round(math.exp(math.log(rr) - z * se), 3),
        "IC_hi": round(math.exp(math.log(rr) + z * se), 3),
    }


def ratio_cotes(a, b, c, d, alpha=0.05):
    """Odds ratio avec IC Wald sur l'échelle log."""
    z = stats.norm.ppf(1 - alpha / 2)
    or_ = (a * d) / (b * c)
    se = math.sqrt(1/a + 1/b + 1/c + 1/d)
    return {
        "OR":    round(or_, 3),
        "IC_lo": round(math.exp(math.log(or_) - z * se), 3),
        "IC_hi": round(math.exp(math.log(or_) + z * se), 3),
    }

# Tabagisme et cancer du poumon (cohorte hypothetique)
# a = fumeurs avec cancer ; b = fumeurs sans cancer
# c = non-fumeurs avec cancer ; d = non-fumeurs sans cancer
print("RR (cohorte)    :", ratio_risque(80, 920, 10, 990))

# Obesite et arthrose du genou (cas-temoin)
print("OR (cas-temoin) :", ratio_cotes(120, 60, 80, 140))
````

## 9. Application — descriptif sur Gapminder

Tableau résumé de l'espérance de vie par continent en 2007.

> **Syntaxe — describe et sélection de colonnes par liste**
> ```python
> df.describe()                          # toutes les colonnes num.
> df.describe(percentiles=[0.25, 0.5, 0.75])  # quartiles
> df_sub = df[["col1", "col2", "col3"]]   # sélection multi-colonnes
> ```

````python
import os
from pathlib import Path
import pandas as pd

# Chemins de cache candidats, dans l'ordre :
#   1. ./data/         (jupyter local, notebook a la racine)
#   2. ../data/        (jupyter local, notebook dans un sous-dossier)
#   3. ./             (au pire on cree un cache dans le cwd)
# En Colab, aucun n'existe : on tombe sur l'URL et on cree ./data/ pour les
# cellules suivantes.
_CANDIDATES = [Path("data"), Path("../data")]
DATA_DIR = next((p for p in _CANDIDATES if p.is_dir()), Path("data"))
DATA_DIR.mkdir(exist_ok=True)


def load_csv(name: str, url: str, **read_csv_kwargs) -> pd.DataFrame:
    """Charge un CSV : cache local d'abord, sinon URL publique.

    Le cache utilise le meme separateur que la lecture (TSV reste TSV).
    """
    for candidate_dir in _CANDIDATES:
        local = candidate_dir / name
        if local.exists():
            return pd.read_csv(local, **read_csv_kwargs)
    df = pd.read_csv(url, **read_csv_kwargs)
    try:
        sep = read_csv_kwargs.get("sep", ",")
        (DATA_DIR / name).parent.mkdir(parents=True, exist_ok=True)
        df.to_csv(DATA_DIR / name, index=False, sep=sep)
    except OSError:
        pass
    return df

GAPMINDER_URL = (
    "https://raw.githubusercontent.com/jennybc/gapminder/"
    "master/inst/extdata/gapminder.tsv"
)
gm = load_csv("gapminder.tsv", GAPMINDER_URL, sep="\t")
recent = gm[gm["year"] == 2007].copy()

# .describe() suit le groupby pour produire un tableau par groupe
resume = (recent.groupby("continent")["lifeExp"]
                .describe(percentiles=[0.25, 0.5, 0.75]).round(1))

# On sélectionné les colonnes utiles
resume[["count", "mean", "std", "min", "25%", "50%", "75%", "max"]]
````

### Exercice 3 — Prevalence avec IC

Une enquete de 2 000 personnes détecté 250 cas. Calculez la prévalence et son IC 95%. Ecrivez une fonction reutilisable.

````python
# Votre essai ici
````

````python
# Solution
def prevalence_ic(cas, total, confiance=0.95):
    """Prevalence et IC normal a un niveau de confiance donne."""
    p = cas / total
    # stats.norm.ppf : quantile de la loi normale
    z = stats.norm.ppf(0.5 + confiance / 2)
    se = math.sqrt(p * (1 - p) / total)
    return {
        "prévalence": round(p, 4),
        "IC_lo":      round(max(0.0, p - z * se), 4),   # borne >= 0
        "IC_hi":      round(min(1.0, p + z * se), 4),   # borne <= 1
    }

print(prevalence_ic(250, 2000))
````
