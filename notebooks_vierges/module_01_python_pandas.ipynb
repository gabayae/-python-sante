## 1. Variables et types

Une **variable** est un nom qui pointe vers une valeur.

> **Syntaxe — affectation**
> ```python
> nom_variable = valeur
> ```
> Le `=` se lit *"reçoit"*. Python **devine** le type a partir de la valeur : `45` -> int, `37.2` -> float, `"texte"` -> str, `True`/`False` -> bool.

| Type    | Stocke                      | Exemple en santé                   |
|---------|-----------------------------|------------------------------------|
| `int`   | Nombre entier               | Age en annees, nombre d'admissions |
| `float` | Nombre a virgule            | Temperature, HbA1c, créatinine     |
| `str`   | Texte                       | Code CIM, identifiant patient      |
| `bool`  | Vrai / Faux                 | Hospitalise ? Fumeur ?             |

> **Syntaxe — f-string (chaine formattee)**
> ```python
> f"texte {variable} texte"          # affiche la valeur de `variable`
> f"valeur = {x:.1f}"                # avec format : 1 chiffre après la virgule
> ```
> Le `f` devant les guillemets active l'interpolation. C'est la facon **recommandée** en Python 3.6+.

````python
# Quatre variables, une par type
age = 45                           # int  - entier
temperature = 37.2                 # float - décimal
diagnostic = "Diabete de type 2"  # str   - texte entre guillemets
hospitalise = True                 # bool  - True ou False (majuscules !)

# print() affiche sur la sortie standard.
# f"...{var}..." insère la valeur de var dans la chaine.
print(f"Age          : {age} ans")
print(f"Temperature  : {temperature} C")
print(f"Diagnostic   : {diagnostic}")
print(f"Hospitalise  : {hospitalise}")

# type(x) renvoie le type de x - utile pour deboguer
print(f"\ntype(age)   = {type(age).__name__}")
print(f"type(temp.) = {type(temperature).__name__}")
````

## 2. Listes — plusieurs mesures du même type

Une **liste** range des valeurs **dans l'ordre**.

> **Syntaxe — liste**
> ```python
> ma_liste = [valeur1, valeur2, valeur3]   # crochets, virgules
> ma_liste[0]                              # premier  element (indice 0 !)
> ma_liste[-1]                             # dernier  element
> ma_liste[1:3]                            # tranche  : indices 1 et 2
> len(ma_liste)                            # nombre d'elements
> ma_liste.append(valeur)                  # ajoute a la fin (modifie en place)
> ```
> **Attention** : Python compte **a partir de 0**, pas de 1. Le premier element est `[0]`, le deuxieme `[1]`, etc.

````python
# Cinq mesures de tension systolique pour un même patient (mmHg)
tensions = [120, 135, 128, 142, 118]   # crochets, virgules

# Acces aux elements individuels
print(f"Toutes : {tensions}")
print(f"1ere   : {tensions[0]} mmHg")          # indice 0 = premier
print(f"3eme   : {tensions[2]} mmHg")          # indice 2 = troisieme
print(f"Dern.  : {tensions[-1]} mmHg")         # -1 = derniere
print(f"3 dern.: {tensions[-3:]}")             # tranche depuis -3

# Fonctions integrees utiles
print(f"\nNombre : {len(tensions)}")
print(f"Min    : {min(tensions)} mmHg")
print(f"Max    : {max(tensions)} mmHg")
print(f"Somme  : {sum(tensions)}")
print(f"Moyenne: {sum(tensions) / len(tensions):.1f} mmHg")

# Ajouter une mesure - modifie la liste en place
tensions.append(124)
print(f"\nApres append(124) : {tensions}")
````

## 3. Dictionnaires — la fiche patient

Un **dictionnaire** associe des **clés** (souvent du texte) a des **valeurs**.

> **Syntaxe — dictionnaire**
> ```python
> mon_dict = {
>     "cle1": valeur1,        # paires cle : valeur
>     "cle2": valeur2,        # separees par des virgules
> }
> mon_dict["cle1"]              # lecture
> mon_dict["cle3"] = valeur3    # ajout / mise a jour
> "cle1" in mon_dict            # test d'appartenance -> True/False
> mon_dict.keys()               # toutes les clés
> ```
> Une **fiche patient** est naturellement un dictionnaire : chaque champ (`age`, `sexe`, `diagnostic`) est une cle.

````python
# Une fiche patient sous forme de dictionnaire
patient = {
    "id":              "PAT-0042",
    "age":             58,
    "sexe":            "F",
    "diagnostic":      "HTA",                     # hypertension
    "tension_systolique": 148,
    "medicaments":     ["Amlodipine", "HCTZ"],   # liste imbriquee
    "fumeur":          False,
}

# Lecture d'un champ
print(f"Patient {patient['id']} : {patient['age']} ans, {patient['sexe']}")
print(f"Diagnostic : {patient['diagnostic']}")

# La clé 'medicaments' contient une liste - on peut l'iterer
print(f"Medicaments ({len(patient['medicaments'])}) : {', '.join(patient['medicaments'])}")

# Mise a jour d'un champ - utile pour les visites de suivi
patient["tension_systolique"] = 138
print(f"Tension après traitement : {patient['tension_systolique']} mmHg")

# Test d'appartenance
print(f"\nLe patient a-t-il un champ 'allergies' ? {'allergies' in patient}")
````

## 4. Conditionnelles — appliquer une règle clinique

> **Syntaxe — if / elif / else**
> ```python
> if condition1:
>     # bloc exécute si condition1 est True
>     ...
> elif condition2:
>     # bloc exécute si condition1 False ET condition2 True
>     ...
> else:
>     # bloc exécute si toutes les précédentes sont False
>     ...
> ```
> **Indentation** : Python utilise des **espaces** (4 par convention) pour delimiter les blocs. Pas d'accolades ! Une indentation incorrecte leve une `IndentationError`.

> **Operateurs de comparaison** : `==` (égal), `!=` (different), `<`, `<=`, `>`, `>=`.  
> **Operateurs booleens** : `and` (et), `or` (ou), `not` (non).

````python
# Classification ACC/AHA 2017 de l'hypertension
systolique = 148  # mmHg

# L'ordre des conditions compte : on teste du seuil le plus haut
# au plus bas. Sinon on classerait 148 comme "Elevee" au lieu de "Stade 2".
if systolique >= 140:
    categorie = "Stade 2 Hypertension"
elif systolique >= 130:
    categorie = "Stade 1 Hypertension"
elif systolique >= 120:
    categorie = "Elevee"
else:
    categorie = "Normale"

print(f"Tension {systolique} mmHg  ->  {categorie}")
````

## 5. Boucles `for` — appliquer une règle a toute la cohorte

> **Syntaxe — for**
> ```python
> for element in collection:        # collection = liste, dict, ...
>     # corps de la boucle, exécute une fois par element
>     print(element)
> ```
> Le `for` parcourt **dans l'ordre** chaque element. La variable de boucle (`element` ci-dessus) prend successivement chaque valeur.

Combinaison fréquente : on a une cohorte (liste de dictionnaires) et une règle clinique a appliquer a chaque patient.

Exemple : depistage du diabète par HbA1c (critères ADA) :

- HbA1c < 5,7 % -> normal
- 5,7 a 6,4 % -> pre-diabète
- >= 6,5 % -> diabète

````python
# Cinq patients sous forme de liste de dictionnaires
cohorte = [
    {"id": "P001", "age": 34, "hba1c": 5.4},
    {"id": "P002", "age": 67, "hba1c": 7.8},
    {"id": "P003", "age": 52, "hba1c": 6.1},
    {"id": "P004", "age": 29, "hba1c": 5.0},
    {"id": "P005", "age": 73, "hba1c": 9.1},
]

# La variable `p` reçoit successivement chaque dictionnaire de la liste
for p in cohorte:
    # On peut combiner if/elif/else dans la boucle
    if p["hba1c"] >= 6.5:
        statut = "Diabetique"
    elif p["hba1c"] >= 5.7:
        statut = "Pre-diabétique"
    else:
        statut = "Normal"
    # Affichage formate : :>2d aligne les chiffres a droite sur 2 caractères,
    # :>4.1f reserve 4 caractères avec 1 chiffre après la virgule
    print(f"  {p['id']} (age {p['age']:>2d}) : HbA1c = {p['hba1c']:>4.1f}%  ->  {statut}")
````

## 6. Fonctions — emballer une formule clinique

Une **fonction** est un bloc de code reutilisable, nomme, avec des parametres en entree et éventuellement une valeur en sortie.

> **Syntaxe — fonction**
> ```python
> def nom_fonction(param1, param2):       # signature
>     """Docstring : une ligne qui dit ce que fait la fonction."""
>     # corps de la fonction
>     ...
>     return résultat                     # renvoie une valeur
>
> # Appel :
> r = nom_fonction(valeur1, valeur2)      # positionnel
> r = nom_fonction(param1=valeur1)        # par mot-cle (plus lisible !)
> ```
> **Annotations de type** (optionnelles mais recommandées) :
> ```python
> def imc(poids_kg: float, taille_m: float) -> float:
>     ...
> ```
> Elles documentent les types attendus sans rien forcer a l'exécution.

> **Docstring** : la chaine entre triple guillemets juste après `def`. Affichee par `help(nom_fonction)` ou par Jupyter quand on survole le nom.

````python
def imc(poids_kg: float, taille_m: float) -> float:
    """Indice de Masse Corporelle en kg/m^2.

    Parametres
    ----------
    poids_kg : float
        poids en kilogrammes.
    taille_m : float
        taille en metres (PAS en cm).
    """
    # Verification defensive : taille positive
    if taille_m <= 0:
        raise ValueError("taille_m doit être strictement positive")
    # ** est l'opérateur puissance (taille^2)
    return poids_kg / (taille_m ** 2)


def categorie_imc(valeur: float) -> str:
    """Classification OMS adulte."""
    if valeur < 18.5:
        return "Maigreur"
    if valeur < 25.0:
        return "Normal"
    if valeur < 30.0:
        return "Surpoids"
    return "Obesite"

# Appel par mots-clés - plus lisible que imc(82, 1.75)
v = imc(poids_kg=82, taille_m=1.75)
print(f"IMC = {v:.1f}  ({categorie_imc(v)})")

# La docstring est accessible avec help() ou .__doc__
print(f"\nDocstring : {imc.__doc__.splitlines()[0]}")
````

## 7. Pandas — du Python pur a la table

**Pandas** est la bibliothèque standard pour manipuler des tableaux en Python. L'objet central est le **`DataFrame`** : un tableur programmable.

> **Syntaxe — importer pandas et charger un CSV**
> ```python
> import pandas as pd                              # convention
>
> df = pd.read_csv("fichier.csv")                  # depuis le disque
> df = pd.read_csv("https://...../fichier.csv")    # depuis Internet
> df = pd.read_csv("fichier.tsv", sep="\t")       # avec un autre separateur
> ```

On charge ici **Gapminder** : espérance de vie, PIB par habitant et population pour ~140 pays tous les 5 ans, de 1952 a 2007. [GitHub : `jennybc/gapminder`](https://github.com/jennybc/gapminder).

La fonction `load_csv` ci-dessous est notre **patron de chargement** utilise dans tous les modules. Elle essaie le cache local (`data/`) d'abord, sinon telecharge depuis l'URL. Cela marche **en Colab comme en local**.

````python
import os
from pathlib import Path
import pandas as pd

# Chemins de cache candidats, dans l'ordre :
#   1. ./data/         (jupyter local, notebook a la racine)
#   2. ../data/        (jupyter local, notebook dans un sous-dossier)
#   3. ./             (au pire on cree un cache dans le cwd)
# En Colab, aucun n'existe : on tombe sur l'URL et on cree ./data/ pour les
# cellules suivantes.
_CANDIDATES = [Path("data"), Path("../data")]
DATA_DIR = next((p for p in _CANDIDATES if p.is_dir()), Path("data"))
DATA_DIR.mkdir(exist_ok=True)


def load_csv(name: str, url: str, **read_csv_kwargs) -> pd.DataFrame:
    """Charge un CSV : cache local d'abord, sinon URL publique.

    Le cache utilise le meme separateur que la lecture (TSV reste TSV).
    """
    for candidate_dir in _CANDIDATES:
        local = candidate_dir / name
        if local.exists():
            return pd.read_csv(local, **read_csv_kwargs)
    df = pd.read_csv(url, **read_csv_kwargs)
    try:
        sep = read_csv_kwargs.get("sep", ",")
        (DATA_DIR / name).parent.mkdir(parents=True, exist_ok=True)
        df.to_csv(DATA_DIR / name, index=False, sep=sep)
    except OSError:
        pass
    return df

# URL stable de Gapminder (depuis le repo de Jenny Bryan)
GAPMINDER_URL = (
    "https://raw.githubusercontent.com/jennybc/gapminder/"
    "master/inst/extdata/gapminder.tsv"
)

import pandas as pd                       # convention : pd

# sep="\t" parce que c'est un TSV (tab-separated values)
gm = load_csv("gapminder.tsv", GAPMINDER_URL, sep="\t")

# .shape renvoie un tuple (lignes, colonnes)
print(f"Dimensions : {gm.shape[0]:,} lignes x {gm.shape[1]} colonnes")

# .head() affiche les n premières lignes (défaut 5)
gm.head()
````

## 9. Filtrer des lignes (masque booleen)

Le filtrage utilise un **masque booleen** : une colonne de `True`/`False` qu'on passe entre crochets.

> **Syntaxe — filtrage**
> ```python
> masque = df["col"] > seuil          # cree une Series de True/False
> sous_df = df[masque]                # garde les lignes ou masque == True
>
> # Ou en une ligne :
> sous_df = df[df["col"] > seuil]
>
> # Combiner avec ET (&), OU (|), NON (~)
> sous_df = df[(df["a"] > 1) & (df["b"] == "X")]
> ```
> **Attention** : il faut `&` `|` `~` (pas `and` `or` `not`) et **parentheser chaque condition** — sinon precedence des opérateurs vous trahira.

````python
# La derniere année présente dans le jeu de données
annee_recente = gm["year"].max()       # .max() renvoie le max
print(f"Annee la plus recente : {annee_recente}")

# Tous les pays pour l'année la plus récente
# (le == compare valeur par valeur ; le résultat est une Series booleenne)
recent = gm[gm["year"] == annee_recente]
print(f"Nb pays cette annee   : {len(recent)}")

# Sous-filtrage : continent africain dans cette même année
afrique_recent = recent[recent["continent"] == "Africa"]
print(f"Pays africains        : {len(afrique_recent)}")
afrique_recent.head()
````

````python
# Esperance de vie < 50 ans dans la derniere année
faible_ev = recent[recent["lifeExp"] < 50]
print(f"Pays avec EV < 50 ans en {annee_recente} : {len(faible_ev)}")

# Selection multi-colonnes : double crochet puisque c'est une liste
# .sort_values(col) trie selon une colonne
faible_ev[["country", "continent", "lifeExp"]].sort_values("lifeExp")
````

## 11. `groupby` + `agg` — le geste pivot de pandas

Le **patron split-apply-combine** :

1. **Split** : couper les lignes en groupes selon une categorie.
2. **Apply** : calculer une statistique sur chaque groupe.
3. **Combine** : recoller le tout en une table.

> **Syntaxe — groupby**
> ```python
> df.groupby("col_categ")["col_num"].mean()             # une fonction
> df.groupby("col_categ")["col_num"].agg(["mean", "std"])   # plusieurs
> df.groupby(["c1", "c2"])["x"].mean()                  # par croisement
> ```
> `.agg()` accepte une liste de noms de fonctions ou même une fonction personnalisee.

````python
# Statistiques d'EV par continent en 2007
# Methode chainee : on lit de gauche a droite comme une phrase
resume = (
    recent                                                # le DataFrame
    .groupby("continent")["lifeExp"]                     # par continent, colonne EV
    .agg(["mean", "median", "std", "min", "max", "count"])  # 6 stats
    .round(1)                                              # arrondi cosmetique
    .sort_values("mean", ascending=False)                  # tri
)
resume
````

## 12. Creer une colonne — `pd.cut` pour catégoriser

On veut souvent transformer une variable **continue** (EV en annees) en catégorie **ordonnee** (Faible / Moyen / Eleve).

> **Syntaxe — pd.cut**
> ```python
> df["categ"] = pd.cut(
>     df["col_continue"],
>     bins=[seuil0, seuil1, seuil2, seuil3],   # 4 bornes -> 3 catégories
>     labels=["basse", "moyenne", "haute"],   # noms des catégories
> )
> ```
> Les bornes sont **incluses a droite** par défaut (`right=True`).

````python
# .copy() empeche le SettingWithCopyWarning quand on modifie
gm = gm.copy()

# 4 bornes -> 3 intervalles -> 3 labels
# [0, 50] -> "Tres faible", (50, 65] -> "Faible", etc.
gm["categorie_ev"] = pd.cut(
    gm["lifeExp"],
    bins=[0, 50, 65, 75, 100],
    labels=["Tres faible (<50)", "Faible (50-65)",
            "Moyen (65-75)", "Eleve (>=75)"],
)

# .sample(n, random_state=...) tire n lignes au hasard, de manière reproductible
gm[["country", "year", "lifeExp", "categorie_ev"]].sample(5, random_state=0)
````

> ```python
> pd.crosstab(df["col_ligne"], df["col_colonne"])   # comptes
